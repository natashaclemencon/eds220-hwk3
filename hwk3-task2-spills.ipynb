````python
# FINAL CODE

# Import libraries.
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd

# Read in datasets. 
spills = pd.read_csv('data/Spill_Incidents_20261103.csv')
counties = gpd.read_file('data/tl_2026_us_county (1)/tl_2026_us_county.shp')

# Clean spills data.
# Change the column names in spills to all be lower case, to have no spaces before or after
# the column name, and to have the remaining spaces be replaced with a underscore. 
spills.columns = spills.columns.str.lower().str.strip().str.replace(" ","_")
# Change the date columns to datetime objects. 
spills['spill_date']=pd.to_datetime(spills['spill_date'])
spills['received_date']=pd.to_datetime(spills['received_date'])
spills['close_date']=pd.to_datetime(spills['close_date'])

# Create petr_23 dataframe by filtering spills between January 1 and December 31 2023, and 
# for where material name is equal to Petroleum. 
petr_23 = spills[(spills['spill_date']>='2023-01-01') & (spills['spill_date']<='2023-12-31') & (spills['material_family']=="Petroleum")]

# Use petr_23 to create a new df with the number of petroleum spills per county in 2023.
spills_per_county = petr_23.groupby('county')['spill_number'].nunique()
# Reset the index.
spills_per_county = spills_per_county.reset_index()
# Rename spill_number column to n_spills.
spills_per_county = spills_per_county.rename(columns={"spill_number":"n_spills"})

# Clean counties data.
# Make all the columns in counties lowercase, and remove any excess spaces. 
counties.columns = counties.columns.str.lower().str.strip()

# Create county variable of just counties in New York. 
ny_counties = counties[counties['statefp']=='36']

# Change the value "St Lawrence" to "St. Lawrence" in the spills_per_county county column.
spills_per_county['county']= spills_per_county['county'].replace("St Lawrence", "St. Lawrence")
# Use an assert to make sure the change went through. 

# Change the name column in ny_counties to 'county' so it can be merged with spills_per_county on 'county'.
ny_counties = ny_counties.rename(columns={"name":"county"})

# Merge spills_per_county and ny_counties.
ny_spills = ny_counties.merge(spills_per_county, how='inner', on='county')

# Add Indigenous Reservation spill numbers to existing counties. 
# Add a value of 2 to the Niagara county. 
ny_spills.loc[ny_spills['county']=='Niagara', 'n_spills']+=2
# Add a value of 1 to the Onondaga county. 
ny_spills.loc[ny_spills['county']=='Onondaga', 'n_spills']+=1

# Plot spills per county in New York in 2023. 
fig, ax = plt.subplots(figsize=(8,6))

ax.axis('off')

ny_spills.plot(
    ax=ax,
    column='n_spills',
    legend=True,
    cmap='Blues', legend_kwds={
                          "label":"Number of Spills \n accessed Nov 2026 from (New York State Department of Environmental Conservation)",
                          "orientation": "horizontal"}
        )
ax.set_title('Number of Petroleum Spills in New York Counties in 2023')

plt.show()


````
